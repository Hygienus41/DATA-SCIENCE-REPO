Why?

k = 1

Very sensitive to noise. Often overfits the training data. Can perform worse on unseen test data.

k = 5

Uses 5 neighbors to vote on the class. Smooths out noise. Usually generalizes better, giving higher test accuracy.

Intuition: Imagine recognizing a handwritten digit:

k = 1: You trust just one neighbor → risky if that neighbor is mislabeled. k = 5: You ask five nearby digits to vote → more reliable.

# Train and validation errors initialized as empty list
train_errs = list()
valid_errs = list()

C_values = [0.001, 0.01, 0.1, 1, 10, 100, 1000]

# Loop over values of C_value
for C_value in C_values:
    # Create LogisticRegression object and fit
    lr = LogisticRegression(C=C_value, max_iter=1000)
    lr.fit(X_train, y_train)
    
    # Evaluate error rates and append to lists
    train_errs.append(1.0 - lr.score(X_train, y_train))
    valid_errs.append(1.0 - lr.score(X_valid, y_valid))
    
# Plot results
plt.semilogx(C_values, train_errs, C_values, valid_errs)
plt.legend(("train", "validation"))
plt.show()


### When you look at the plot: Best C = where validation error is lowest, Not training error.

🎯 What you’ll likely see 

Training error ↓ as C increases
Validation error:
first ↓ (model improves) then ↑ (overfitting starts)

👉 So the curve looks like a U-shape

🔥 Real-world intuition

Too simple → misses patterns ❌, Too complex → memorizes noise ❌
Just right → generalizes well ✅
