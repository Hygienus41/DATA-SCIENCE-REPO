```python
# Remove leading and trailing spaces from all column names
df.columns = df.columns.str.strip()
```

```python
# Missing Values Heatmap

plt.figure(figsize=(10,5))

sns.heatmap(
    df.isnull(),
    cbar=False,
    yticklabels=False
)

plt.title("Missing Values Heatmap")
plt.xlabel("Columns")
plt.show()
```

```python
# For every numerical column, group the data by country. Inside each country, find missing values and replace them with that country's median value. Put the cleaned values back into the dataframe."
num_cols = df.select_dtypes(include='number').columns

for col in num_cols:
    df[col] = df.groupby('Country')[col].transform(
        lambda x: x.fillna(x.median())
    )

for col in num_cols:
    df[col] = df[col].fillna(df[col].median())
```

```python
# confirm if there are still missing values
df.isnull().sum().sum()
```

```python
# Duplicate Analysis

df.duplicated().sum()
# df[df.duplicated()]
```

```python
# Outlier Detection

numeric_columns = df.select_dtypes(include='number').columns

for col in numeric_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower) | (df[col] > upper)]

    print(f"\nOutliers in {col}:")
    print(outliers)
```

# Does GDP Affect Life Expectancy?

This analysis investigates whether countries with higher GDP per capita tend to have higher life expectancy.

```python
# Check the correlation

correlation = df['GDP'].corr(df['Life expectancy'])

print(f"Correlation coefficient: {correlation:.2f}")

plt.figure(figsize=(8,6))

plt.scatter(df['GDP'], df['Life expectancy'])

plt.title("GDP vs Life Expectancy")
plt.xlabel("GDP per Capita")
plt.ylabel("Life Expectancy (Years)")

plt.show()
```

There is a moderate positive correlation between GDP per capita and life expectancy, suggesting that wealthier countries generally have longer life expectancy.

### One issue with GDP is that it is usually highly skewed—a few very wealthy countries stretch the x-axis and compress most of the data. A common improvement is to plot the logarithm of GDP instead:

```python
plt.figure(figsize=(8,6))

plt.scatter(np.log10(df['GDP']), df['Life expectancy'])

plt.title("Life Expectancy vs Log(GDP)")
plt.xlabel("Log10(GDP per Capita)")
plt.ylabel("Life Expectancy (Years)")

plt.show()
```

# Healthy overperformers (Countries with High Life Expectancy Despite Low GDP)

```python
high_life_low_gdp = df[
    (df['GDP'] < df['GDP'].quantile(0.25)) &
    (df['Life expectancy'] > df['Life expectancy'].quantile(0.75))
]

plt.figure(figsize=(10, 6))

# Plot all countries
plt.scatter(
    df['GDP'],
    df['Life expectancy'],
    color='lightgray',
    alpha=0.6,
    label='Other Countries'
)

# Highlight outliers
plt.scatter(
    high_life_low_gdp['GDP'],
    high_life_low_gdp['Life expectancy'],
    color='red',
    label='Health Overperformers'
)

# Label the outlier countries
for _, row in high_life_low_gdp.iterrows():
    plt.text(
        row['GDP']*1.02,
        row['Life expectancy'],
        row['Country'],
        fontsize=8,
        ha='left',
        va='bottom'
    )

plt.title("Countries with High Life Expectancy Despite Low GDP")
plt.xlabel("GDP per Capita")
plt.ylabel("Life Expectancy (Years)")
plt.legend()
plt.xscale('log')

plt.show()
```

### Insight

The scatter plot shows a positive relationship between GDP per capita and life expectancy. Countries with higher GDP generally tend to have longer life expectancy, although the relationship is not perfectly linear. Several countries achieve relatively high life expectancy despite lower GDP (Healthy overperformers) levels, suggesting that factors such as effective healthcare systems, education, public health policies, and healthy lifestyles also play an important role.

# Adult Mortality vs Life Expectancy

This analysis explores the relationship between adult mortality and life expectancy. Since adult mortality reflects the likelihood of dying between the ages of 15 and 60, we expect countries with higher adult mortality to have lower average life expectancy.

```python
# Correlation Analysis

correlation = df['Adult Mortality'].corr(df['Life expectancy'])

print(f"Correlation coefficient: {correlation:.2f}")

# Scatter Plot

plt.figure(figsize=(8,6))

plt.scatter(df['Adult Mortality'], df['Life expectancy'])

plt.title("Adult Mortality vs Life Expectancy")
plt.xlabel("Adult Mortality")
plt.ylabel("Life Expectancy (Years)")

plt.show()
```

## Average adult mortality for each country

```python
# Calculate the average adult mortality for each country:

country_mortality = (
    df.groupby('Country')[['Adult Mortality', 'Life expectancy']]
      .mean()
      .reset_index()
)

# Select the top 20:

top20 = country_mortality.nlargest(20, 'Adult Mortality')

top20[['Country', 'Adult Mortality', 'Life expectancy']].head()
```

```python
plt.figure(figsize=(10, 6))

# Plot all countries
plt.scatter(
    country_mortality['Adult Mortality'],
    country_mortality['Life expectancy'],
    color='lightgray',
    alpha=0.5,
    label='All Countries'
)

# Highlight the top 20 countries
plt.scatter(
    top20['Adult Mortality'],
    top20['Life expectancy'],
    color='red',
    s=80,                    # Larger marker size
    edgecolors='black',      # Black border around markers
    label='Top 20 Highest Adult Mortality'
)

# Add country labels
for x, y, country in zip(
    top20['Adult Mortality'],
    top20['Life expectancy'],
    top20['Country']
):
    plt.text(
        x + 5,
        y + 0.3,
        country,
        fontsize=8
    )

# Titles and labels
plt.title("Countries with the Highest Adult Mortality", fontsize=14)
plt.xlabel("Average Adult Mortality", fontsize=12)
plt.ylabel("Average Life Expectancy (Years)", fontsize=12)

# Add grid
plt.grid(alpha=0.3)

# Add legend
plt.legend()

# Make the layout neater
plt.tight_layout()

plt.show()
```

### Insight

The analysis reveals a strong negative relationship between adult mortality and life expectancy. Countries with higher adult mortality generally experience substantially lower life expectancy. This suggests that reducing premature adult deaths through improved healthcare, disease prevention, and access to medical services could have a significant impact on increasing overall life expectancy.

# Does Health Expenditure Affect Life Expectancy?

This analysis investigates whether countries that spend more on healthcare tend to have higher life expectancy. It also identifies countries that spend heavily but achieve only moderate health outcomes, as well as countries that achieve long life expectancy with relatively low health expenditure.

```python
# Correlation Analysis
health_corr = df['Total expenditure'].corr(df['Life expectancy'])

print(f"Correlation coefficient: {health_corr:.2f}")


# Scatter Plot

plt.figure(figsize=(8,6))

plt.scatter(
    df['Total expenditure'],
    df['Life expectancy'],
    alpha=0.5
)

plt.title("Health Expenditure vs Life Expectancy")
plt.xlabel("Health Expenditure (%)")
plt.ylabel("Life Expectancy (Years)")

plt.show()
```

Interpretation

The correlation coefficient of 0.23 indicates a weak positive relationship between health expenditure and life expectancy. Countries that spend more on healthcare tend to have slightly higher life expectancy, but the relationship is not strong. This suggests that increasing healthcare expenditure alone does not necessarily lead to longer lives. Other factors such as education, economic development, disease burden, healthcare efficiency, sanitation, and access to quality medical services also play significant roles in determining life expectancy.

```python
# Countries Spending Heavily but Achieving Mediocre Outcomes

country_health = (
    df.groupby('Country')[['Total expenditure', 'Life expectancy']]
      .mean()
      .reset_index()
)

country_health.head()
```

Note: The Total expenditure variable represents the percentage of total government expenditure allocated to health, not the absolute amount of money spent. Therefore, countries with similar expenditure percentages may still differ substantially in healthcare quality, efficiency, and available resources, which may explain the relatively weak correlation with life expectancy.

```python
# Identify countries that Spend above the 75th percentile and have below the median life expectancy

high_spending = country_health['Total expenditure'].quantile(0.75)

median_life = country_health['Life expectancy'].median()

inefficient = country_health[
    (country_health['Total expenditure'] >= high_spending) &
    (country_health['Life expectancy'] < median_life)
]

inefficient = inefficient.sort_values('Total expenditure')

print(inefficient)
```

```python
# Countries Spending Efficiently.These countries achieve relatively good outcomes despite lower health expenditure.

efficient = country_health[
    (country_health['Total expenditure'] < country_health['Total expenditure'].median()) &
    (country_health['Life expectancy'] > median_life)
]

efficient = efficient.sort_values('Total expenditure')

print(efficient.head(15))
```

```python
# Visualize Efficient Countries

plt.figure(figsize=(10,6))

# All countries
plt.scatter(
    country_health['Total expenditure'],
    country_health['Life expectancy'],
    color='lightgray',
    alpha=0.5
)

# Efficient countries
plt.scatter(
    efficient['Total expenditure'],
    efficient['Life expectancy'],
    color='green',
    s=80,
    edgecolors='black'
)

for x, y, country in zip(
    efficient['Total expenditure'],
    efficient['Life expectancy'],
    efficient['Country']
):
    plt.text(x + 0.1, y + 0.2, country, fontsize=8)

plt.title("Countries Achieving High Life Expectancy with Lower Health Expenditure")
plt.xlabel("Average Health Expenditure (%)")
plt.ylabel("Average Life Expectancy (Years)")

plt.grid(alpha=0.3)

plt.show()
```

```python
# Comparing the relationship between GDP and Total_expenditure on life_expectancy

print(f"GDP correlation: {gdp_corr:.2f}")
print(f"Health expenditure correlation: {health_corr:.2f}")
```

### Interpretation

GDP has a stronger positive relationship with life expectancy (0.43) than health expenditure (0.23). While wealthier countries generally tend to have longer life expectancy, allocating a larger percentage of government expenditure to health shows only a weak association with lifespan. This suggests that simply increasing healthcare spending does not guarantee better health outcomes. The effectiveness of healthcare systems, education, disease prevention, sanitation, nutrition, and access to quality medical services are also important determinants of life expectancy.

### Business/Healthcare Insight

This finding implies that resource allocation and efficiency may matter more than expenditure alone. Two countries may devote a similar percentage of their government budget to healthcare, yet achieve very different life expectancy outcomes because of differences in healthcare quality, infrastructure, policy implementation, and broader social conditions.

The correlation coefficient of 0.40 indicates a moderate positive relationship between alcohol consumption and life expectancy. Instead, the relationship is likely influenced by confounding factors such as higher income, better healthcare systems, improved education, and higher standards of living, which are common in many developed countries.

Healthcare Insight

Alcohol consumption alone is not a reliable predictor of population health. Although the analysis shows a moderate positive association, it is likely that socioeconomic development and healthcare quality explain much of the observed relationship. ## Baseline Multiple Linear Regression Model

A multiple linear regression model was developed to predict **life expectancy** using GDP, schooling, BMI, adult mortality, and HIV/AIDS prevalence. This model serves as a baseline for estimating expected life expectancy and identifying countries that perform better or worse than predicted based on these key health and socioeconomic indicators.

```python
# Build a Baseline Multiple Linear Regression Model

from sklearn.linear_model import LinearRegression

X =df[[
    'GDP',
    'Schooling',
    'BMI',
    'Adult Mortality',
    'HIV/AIDS'
]]
y = df['Life expectancy']

model = LinearRegression()
model.fit(X, y)

# Predict Life Expectancy
df['Predicted Life Expectancy'] = model.predict(X)
df.head()
```

```python
# Calculate the Difference
# This is the key metric. Positive → Better than expected, Negative → Worse than expected

df['Difference'] = (
    df['Life expectancy']
    - df['Predicted Life Expectancy']
)

df.head()
```

```python
# Average by Country (Because each country appears multiple times (different years), summarize by country.)

country_performance = (
    df.groupby('Country')[['Difference']]
      .mean()
      .reset_index()
)

country_performance
```

```python
# Best Performing Countries. These countries have higher life expectancy than their GDP would predict.
best = country_performance.nlargest(10, 'Difference')
print(best)

# Visualize the Best Performers

plt.figure(figsize=(10,6))

bars = plt.barh(
    best['Country'],
    best['Difference']
)

for bar in bars:
    plt.text(
        bar.get_width(),
        bar.get_y() + bar.get_height()/2,
        f"{bar.get_width():.2f}",
        va='center'
    )

plt.title("Countries Performing Better Than Expected")
plt.xlabel("Years Above Expected Life Expectancy")

plt.grid(axis='x', alpha=0.3)

plt.tight_layout()

plt.show()
```

```python
# Worst Performing Countries. These countries have lower life expectancy than expected given their GDP.
worst = country_performance.nsmallest(10, 'Difference')

print(worst)

# Visualize the Worst Performers
plt.figure(figsize=(10,6))

bars = plt.barh(
    worst['Country'],
    worst['Difference']
)

for bar in bars:
    plt.text(
        bar.get_width(),
        bar.get_y() + bar.get_height()/2,
        f"{bar.get_width():.2f}",
        va='center'
    )

plt.title("Countries Performing Worse Than Expected")
plt.xlabel("Years Below Expected Life Expectancy")

plt.grid(axis='x', alpha=0.3)

plt.tight_layout()

plt.show()
```

# Health Inequality Analysis

```python
# Create Country-Level Data. Average the variables for each country first.

country_data = (
    df.groupby('Country')[[
        'Life expectancy',
        'GDP',
        'Schooling'
    ]]
    .mean()
    .reset_index()
)

country_data
```

### Find the Top and Bottom 20%

```python
# Rank countries by life expectancy.

top20 = country_data.nlargest(
    int(len(country_data) * 0.20),
    'Life expectancy'
)

top20.head()
```

```python
bottom20 = country_data.nsmallest(
    int(len(country_data) * 0.20),
    'Life expectancy'
)

bottom20.head()
```

```python
# Compare the Averages

comparison = pd.DataFrame({
    'Indicator': [
        'Life Expectancy',
        'GDP',
        'Schooling'
    ],
    'Top 20%': [
        top20['Life expectancy'].mean(),
        top20['GDP'].mean(),
        top20['Schooling'].mean()
    ],
    'Bottom 20%': [
        bottom20['Life expectancy'].mean(),
        bottom20['GDP'].mean(),
        bottom20['Schooling'].mean()
    ]
}).round(2)

comparison
```

```python
# Calculate the inequality Gaps

comparison['Gap'] = (
    comparison['Top 20%']
    - comparison['Bottom 20%']
).round(2)

comparison
```

## GDP is measured in thousands (or more), while Schooling and Life Expectancy are much smaller numbers. If you plot them on the same y-axis, the GDP bars will dominate the chart and make the other two hard to compare. So a better option is to normalize the indicators before plotting

```python
comparison_plot = comparison.copy()

comparison_plot[['Top 20%', 'Bottom 20%']] = (
    comparison_plot[['Top 20%', 'Bottom 20%']]
    .div(comparison_plot[['Top 20%', 'Bottom 20%']].max(axis=1), axis=0)
)
```

```python
# Visualize

x = np.arange(len(comparison))
width = 0.35

plt.figure(figsize=(8,5))

plt.bar(
    x - width/2,
    comparison_plot['Top 20%'],
    width,
    label='Top 20%'
)

plt.bar(
    x + width/2,
    comparison_plot['Bottom 20%'],
    width,
    label='Bottom 20%'
)

plt.xticks(x, comparison['Indicator'])

plt.title("Health Inequality Analysis")
plt.ylabel("Average Value")
plt.legend()

plt.tight_layout()
plt.show()
```

```python
# Calculate the relative differences between the top and bottom 20% of countries.

comparison['Times Higher'] = (
    comparison['Top 20%'] /
    comparison['Bottom 20%']
).round(2)

comparison['% Difference'] = (
    (comparison['Gap'] / comparison['Bottom 20%']) * 100
).round(2)

comparison
```

### Insight

The health inequality analysis reveals substantial disparities between the top and bottom 20% of countries ranked by life expectancy. The top-performing countries have an average life expectancy that is **25.43 years higher (46.46%)**, an average GDP that is approximately **17 times greater**, and **70.44% more years of schooling** than the bottom-performing countries. These findings demonstrate that countries with stronger economies and higher educational attainment tend to achieve significantly better health outcomes and longer life expectancy.

### Healthcare Insight

The analysis highlights significant global health inequalities. Countries in the top 20% by life expectancy live **25.43 years longer**, have a GDP that is approximately **17 times higher**, and receive about **1.7 times more years of schooling** than countries in the bottom 20%. These findings suggest that better health outcomes are closely linked to economic development and education, emphasizing that reducing health inequalities requires investment in both healthcare and broader social and economic development.
